# Curso Git

## Gui vs cli

### GUI (Graphical User Interface)

Interfaz gráfica con ventanas, botones, menús y elementos visuales.

Ejemplo: GitHub en el navegador, donde puedes crear un fork o abrir un pull request con botones.

### CLI (Command Line Interface)

Interfaz basada en texto donde escribes comandos en la terminal.

Ejemplos: git commit -m "mensaje"
